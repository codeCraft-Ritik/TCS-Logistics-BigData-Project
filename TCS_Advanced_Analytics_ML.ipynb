## 8. PREDICTIVE ANALYTICS APPLICATIONS

```python
print("\n" + "="*80)
print("PREDICTIVE ANALYTICS APPLICATIONS".center(80))
print("="*80)

print("\n1. FUEL CONSUMPTION PREDICTION:")
print("-" * 60)
print("  Application: Predict fuel requirements for new trips")
print(f"  Model Performance: R² = {ml_models.results['fuel_consumption']['r2']:.4f}")
print(f"  Prediction Error: ±{ml_models.results['fuel_consumption']['rmse']:.2f} gallons")

# Example prediction
sample_trip = pd.DataFrame({
    'actual_distance_miles': [1500],
    'actual_duration_hours': [25],
    'idle_time_hours': [5]
})
scaler = ml_models.scalers['fuel_consumption']
sample_scaled = scaler.transform(sample_trip)
predicted_fuel = ml_models.models['fuel_consumption'].predict(sample_scaled)[0]
print(f"\n  Example: For a 1500-mile, 25-hour trip with 5 hours idle time:")
print(f"  → Predicted fuel consumption: {predicted_fuel:.1f} gallons")
print(f"  → Estimated fuel cost: ${predicted_fuel * 3.5:,.2f}")

print("\n2. DELIVERY DELAY PREDICTION:")
print("-" * 60)
if 'delivery_delay' in ml_models.results:
    print(f"  Model Accuracy: {ml_models.results['delivery_delay']['accuracy']:.2%}")
    print("  Application: Identify high-risk trips for delay")
    print("  Recommendation: Proactive customer notification and resource allocation")

print("\n3. ROUTE PROFITABILITY OPTIMIZATION:")
print("-" * 60)
if 'route_profitability' in ml_models.results:
    print(f"  Model R² Score: {ml_models.results['route_profitability']['r2']:.4f}")
    print("  Application: Predict profitability of new routes")
    print("  Recommendation: Dynamic pricing and route assignment strategies")

print("\n4. DRIVER SAFETY RISK ASSESSMENT:")
print("-" * 60)
if 'driver_safety' in ml_models.results:
    print(f"  Model Accuracy: {ml_models.results['driver_safety']['accuracy']:.2%}")
    print("  Application: Identify high-risk drivers for targeted training")
    print("  Recommendation: Safety program investment and monitoring")

print("\n" + "="*80)
```
